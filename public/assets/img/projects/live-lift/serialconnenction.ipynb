```python
import serial
import time

# Set up the serial connection
arduino_port = 'COM4'  # Update this to match your Arduino's port
baud_rate = 9600       # Ensure this matches the Arduino's baud rate

ser = serial.Serial(arduino_port, baud_rate)
time.sleep(2)  # Wait for the Arduino to initialize

goodPosition = True

try:
   if goodPosition:
    ser.write(b'T\n')  # Send the byte string 'T'
    print("Sent: T")

    # Wait to observe the green LED lighting up
    time.sleep(5)  # Adjust the delay as needed
   else: 
    # Send another character to trigger the red LED
    ser.write(b'F')  # Example: Send 'X'
    print("Sent: F")

finally:
    ser.close()
```
